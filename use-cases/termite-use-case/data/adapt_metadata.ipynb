```python
import pandas as pd

# Load your file
df = pd.read_csv("metadata.tsv", sep="\t")

# Extract only King, Queen, Worker from "Casts"
df["Casts"] = df["Casts"].str.extract(r"(King|Queen|Worker)", expand=False)

# Extract colony number from "Casts+colonies"
df["Casts+colonies"] = df["Casts+colonies"].str.extract(r"(Colony\d+)", expand=False)

# Save cleaned file
df.to_csv("cleaned_metadata.tsv", sep="\t", index=False)

df.head()
```
